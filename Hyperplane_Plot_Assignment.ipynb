```python
import numpy as np
import random
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

plt.rc('font', family='serif')
theta_0=3
theta_1=-1
theta_2=1

ax_min=-6
ax_max=6

# Data for plotting
x1 = np.arange(ax_min,ax_max, 1, dtype=int)
#print(x1)
x2 = - (theta_1*x1+theta_0)/theta_2

fig, ax = plt.subplots()
ax.plot(x1, x2)
space = 1
ax.xaxis.set_major_locator(ticker.MultipleLocator(space))
ax.yaxis.set_major_locator(ticker.MultipleLocator(space))

xp_top=np.array([ [3,2], [0.75,-0.75], [-1.5,-1],[1.5,1.5],[3,2],[1.75,3.5],[-1,3] ]).T
xp_bottom=np.array([ [2,-2.75], [3.5,-2], [4,-1] ]).T
ax.plot(xp_top[0],xp_top[1],"+", color="blue")
ax.plot(xp_bottom[0],xp_bottom[1],"_", color="red")

# set x, y-axis limits
ax.set_xlim(ax_min, ax_max)
ax.set_ylim(ax_min, ax_max)

ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

#ax.set(xlabel='x1', ylabel='x2', title='Machine Learning: Hyperplane in 2D')
ax.set_title('Machine Learning: Hyperplane in 2D\n\n')
ax.set_ylabel('x2',loc='top', fontsize=12, color="green", rotation=0, va='bottom')
ax.set_xlabel('x1',loc='right', fontsize=12, color="green")

# Draw and Customized arrow: https://python-fiddle.com/tutorials/matplotlib-arrows
ax.annotate(
    '',
    xy=(1, 1),
    xytext=(2, -0.8),
    arrowprops=dict(
        facecolor='green',
        edgecolor='green',
        linestyle='solid',
        linewidth=2,
        arrowstyle='->',
        shrinkA=0, shrinkB=0
       )
 )

ax.grid(ls=':')
plt.show()
```
