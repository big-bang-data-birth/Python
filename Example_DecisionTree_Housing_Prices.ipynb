```python
'''
https://www.kaggle.com/code/dansbecker/basic-data-exploration/tutorial

Your First Machine Learning Model
'''

import pandas as pd
from google.colab import drive
import os

google_drive_path = '/content/drive/MyDrive/Colab_Notebooks/Data/melb_data.csv' # Replace with your path


# Code to find the current working directory in google drive.
drive.mount('/content/drive')
current_working_directory = os.getcwd()
print(current_working_directory)


# read the data and store data in DataFrame titled melbourne_data
melbourne_data = pd.read_csv(google_drive_path)

# dropna drops missing values (think of na as "not available")
melbourne_data = melbourne_data.dropna(axis=0)


# print all data columns
print(melbourne_data.columns)
# print a summary of the data in Melbourne data
melbourne_data.describe()
```

```python

'''
Prediction Target:  We'll use the dot notation to select the column we want to predict,  which is called the prediction target.
By convention, the prediction target is called y.
'''
y = melbourne_data.Price

'''
Features: The columns that are inputted into our model (and later used to make predictions) are called "features." In our case,
those would be the columns used to determine the home price. Sometimes, you will use all columns except the target as features.
Other times you'll be better off with fewer features. For now, we'll build a model with only a few features.

Later on you'll see how to iterate and compare models built with different features. We select multiple features by providing
a list of column names inside brackets. Each item in that list should be a string (with quotes).
'''
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude']

X = melbourne_data[melbourne_features]

# Get and print dataframe dimensions
rows, columns = X.shape
print(f"X dataframe: rows: {rows}, columns: {columns}")

print(X.describe())    # Display the important data parameters of selected features or columns.
print('\n\n')

print(X.head())               # Display first few rows of the selected data features or columns.

'''
Visually checking your data with these commands is an important part of a data scientist's job.
You'll frequently find surprises in the dataset that deserve further inspection.
'''
```

```python
'''
You will use the scikit-learn library to create your models.
Scikit-learn is easily the most popular library for modeling the types of data typically stored in DataFrames.

  The steps to building and using a model are:
================================================
Define: What type of model will it be?
        - A decision tree
        - Linear Regression
        - Classification
        - Some other type of model? Some other parameters of the model type are specified too.

Fit: Capture patterns from provided data. This is the heart of modeling.

Predict: Just what it sounds like

Evaluate: Determine how accurate the model's predictions are.

Here is an example of defining a decision tree model with scikit-learn and fitting it with the features and target variable.

'''
from sklearn.tree import DecisionTreeRegressor

'''
Define model. Specify a number for random_state to ensure same results on each run
You use any number, and model quality won't depend meaningfully on exactly what value you choose.
'''
melbourne_model = DecisionTreeRegressor(random_state=1)

'''
Fit model
We will now fit the model, so that we can use it to make predictions.
'''
melbourne_model.fit(X, y)
```

```python
'''
In practice, you'll want to make predictions for new houses coming on the market rather than
the houses we already have prices for. But we'll make predictions for the first few rows of
the training data to see how the predict function works.
'''
print("Making predictions for the following 5 houses:")
print(X.head())

print("\n\n The predictions are")
print(melbourne_model.predict(X.head()))
print("\n\nThe actual recorded prices of the houses are")
print(melbourne_data.Price.head())
```
