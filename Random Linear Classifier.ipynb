```python
# Sample code for implementing h(x;th,th0)
#
import numpy as np
# x is dimension d by 1
# th is dimension d by 1
# th0 is dimension 1 by 1 => a scalar or constant
# return 1 by 1 matrix of +1, 0, -1
def positive(x, th, th0):
   return np.sign(np.dot(np.transpose(th), x) + th0)

# Takes a list of numbers and returns a row vector: 1 x n
def rv(value_list):
   return np.array([value_list])

# Takes a list of numbers and returns a column vector:  n x 1
def cv(value_list):
   return np.transpose(rv(value_list))

# 5 data points in two dimensions.
data = np.transpose(np.array([[1, 2], [1, 3], [2, 1], [1, -1], [2, -1]]))
# labels, a 1 by 5 array (1 row, five columns) of 1 and -1 values.
labels = rv([-1, -1, +1, +1, +1])

A = (labels == positive(data, cv([1, 1]), -2))

print("data => ", data)
print("labels", labels)
print("data points on which side of the hyperplane ? =>", A)
```

```python
# Google AI: Random Linear Classifier
import numpy as np
import random
import matplotlib.pyplot as plt

class RandomLinearClassifier:
    def __init__(self, num_features):
        self.weights = np.random.rand(num_features)
        self.bias = np.random.rand()
        print("Weight: ", self.weights, "Bias: ", self.bias)

    def predict(self, features):
      linear_output = np.dot(features, self.weights) + self.bias
      return 1 if linear_output >= 0 else 0

# Generate a list of random integers between a specified range
def generate_random_integers(count, min_val, max_val):
  """Generates a list of random integers.
  Args:
    count: The number of random integers to generate.
    min_val: The minimum value (inclusive).
    max_val: The maximum value (inclusive).

  Returns:
    A list of random integers.
  """
  return [random.randint(min_val, max_val) for _ in range(count)]

# Generate a list of random floats between 0 and 1
def generate_random_floats(count):
  """Generates a list of random floats between 0 and 1.
  Args:
    count: The number of random floats to generate.

  Returns:
    A list of random floats.
  """
  return [random.random() for _ in range(count)]


# Program's Main Entering Point
if __name__ == '__main__':

    # Example usage, generating random numbers (ints, floats)
    num_integers = 10
    min_integer = 1
    max_integer = 5
    random_integers_1 = generate_random_integers(num_integers, min_integer, max_integer)
    print(f"Random integers 1: {random_integers_1}")
    random_integers_2 = generate_random_integers(num_integers, min_integer, max_integer)
    print(f"Random integers 2: {random_integers_2}")

    num_floats = 5
    random_floats = generate_random_floats(num_floats)
    print(f"Random floats: {random_floats}")


    print('\033[4m' + "\n\nRandom Linear Classifier Python Code Example #1 (using dot product): " + '\033[0m')
    #num_features = 4
    num_features = num_integers
    classifier = RandomLinearClassifier(num_features)
    #print("classifier:", classifier)

    #features_1 = np.array([1, 2, 3, 4])
    features_1 = np.array([random_integers_1])
    prediction_1 = classifier.predict(features_1)
    print(f"Prediction_1: {prediction_1}")

    #features_2 = np.array([5, 3, 2, 1])
    features_2 = np.array(random_integers_2)
    prediction_2 = classifier.predict(features_2)
    print(f"Prediction_2: {prediction_2}")


    # https://www.geeksforgeeks.org/python-mean-squared-error/
    print('\033[4m' + "\n\nRandom Linear Classifier Python Code Example #2: (using MSE) " + '\033[0m')

    # Given values
    #Y_true = [1,1,2,2,4]  # Y_true = Y (original values)

    #num_integers= 5
    num_integers= 5000
    min_integer = 1
    max_integer = 5
    Y_true = generate_random_integers(num_integers, min_integer, max_integer)
    #print(f"Y_true: {Y_true}")

    # Calculated values
    #Y_pred = [0.6,1.29,1.99,2.69,3.4]  # Y_pred = Y'
    #Y_pred = generate_random_integers(num_integers, min_integer, max_integer)
    min_range = 1.0
    max_range = 5.0
    Y_pred = np.random.uniform(min_range, max_range,num_integers)
    #print(f"Y_pred: {Y_pred}")

    # Mean Squared Error
    MSE = np.square(np.subtract(Y_true,Y_pred)).mean()

    print("MSE: ", MSE)


#    plt.plot(num_integers, MSE)

#    plt.xlabel("k")
#    plt.ylabel("MSE")
#    plt.title("MSE versus number of samples")

#    plt.show()
```

```python
import random

def generate_hypothesis(num_hypotheses, min_value, max_value):
    """
    Generates a list of randomly generated hypotheses (represented as numerical values)
    and selects the one with the least absolute error.

    Args:
        num_hypotheses (int): The number of hypotheses to generate.
        min_value (int): The minimum possible value for a hypothesis.
        max_value (int): The maximum possible value for a hypothesis.

    Returns:
        tuple: A tuple containing the best hypothesis (numerical value) and its error.
    """
    # Simulate error calculation (replace with your actual error calculation)
    # This is just an example; in a real scenario, you'd have a model or
    # some other method to evaluate each hypothesis's error.
    def calculate_error(hypothesis_value):
        # Example error:  hypothesis value + a random noise
        rand_num=random.uniform(-1,1)
        hypothesis_value_updated=hypothesis_value + rand_num
        abs_hypothesis_value_updated=abs(hypothesis_value_updated)

        print("hypothesis_value=",hypothesis_value, "\thypothesis_value_updated=", hypothesis_value_updated,
              "\t returned_error = abs_hypothesis_value_updated=", abs_hypothesis_value_updated)

        return (abs_hypothesis_value_updated)


    hypotheses = []
    for _ in range(num_hypotheses):
        hypotheses.append(random.randint(min_value, max_value))

    print("hypotheses (randomly generated):", hypotheses)

    # Calculate errors for each hypothesis
    errors = [calculate_error(data_from_hypotheses) for data_from_hypotheses in hypotheses]


    # Find the hypothesis with the least error
    best_hypothesis_index = errors.index(min(errors))       # Store index (location) of least error.
    best_error = errors[best_hypothesis_index]              # Extract the value of least error giving hypotheses.
    best_hypothesis = hypotheses[best_hypothesis_index]     # Extract the value of best hypotheses, which is the index location of least error.

    return best_hypothesis, best_error


if __name__ == "__main__":
    num_hypotheses_to_generate = 10     # default value 10
    min_hypothesis_value = -100         # default value -100
    max_hypothesis_value = 100          # default value 100

    best_hypothesis, best_error = generate_hypothesis(
        num_hypotheses_to_generate, min_hypothesis_value, max_hypothesis_value
    )

    print(f"Best hypothesis: {best_hypothesis}")
    print(f"Best hypothesis error: {best_error}")
```

```python
# Cross Validate Example from W3C

from sklearn import datasets
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import KFold, cross_val_score

X, y = datasets.load_iris(return_X_y=True)

print("X (len)=",X.size/X[0].size,"\nX=",X,"\n y(len)=",y.size,"\n y=",y,)

clf = DecisionTreeClassifier(random_state=42)
print("clf= ",clf)

k_folds = KFold(n_splits = 5)
print("k_folds",k_folds)

scores = cross_val_score(clf, X, y, cv = k_folds)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```
